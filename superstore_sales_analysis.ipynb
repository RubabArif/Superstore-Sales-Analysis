```python
category_analysis = df.groupby('Category')[['Sales','Profit']].sum()
```

```python
category_analysis. sort_values('Profit',ascending= False)
```

```python
import matplotlib.pyplot as plt
```

```python
category_analysis ['Sales'].plot(kind='bar')
plt.title ('Sales by Category')
plt.xlabel('Category')
plt.ylabel('Sales')
plt.xticks(rotation=0)
plt.show()
```

```python
category_analysis['Profit'].plot(kind='bar')
plt.title('Profit by category')
plt.xlabel('category')
plt.ylabel('profit')
plt.xticks(rotation=0)
plt.show()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt. month
```

```python
yearly_sales= df.groupby('Year')['Sales'].sum()
```

```python
yearly_sales.plot(kind = 'bar')
plt.title('Yearly Sales')
plt.xlabel('Year')
plt.ylabel('Sales')
plt.xticks(rotation=0)
plt.show()
```

```python
monthly_sales= df.groupby('Month')['Sales'].sum()
```

```python
monthly_sales.plot(kind='line',marker='o')
plt.title('Month Sales Trend')
plt.xlabel('Months')
plt.ylabel('Sales')
plt.xticks(range(1,13))
plt.show()
```

```python
product_sales = df.groupby('Product Name') ['Sales'].sum()
```

```python
top_10_products = product_sales.sort_values(ascending =False). head(10)
```

```python
top_10_products.sort_values().plot(kind='bar')
plt.title('Top 10 Product by Sales')
plt.xlabel('Sales')
plt.ylabel('Product')
plt.show()
```

```python
product_profit = df.groupby('Product Name')['Profit'].sum()
```

```python
top_10_products=product_profit.sort_values(ascending = False).head(10)
```

```python
top_10_products.sort_values().plot(kind='bar')
plt.title('Top 10 Product by Profit')
plt.xlabel('Profit')
plt.ylabel('Product')
plt.show()
```

```python
loss_products= product_profit[product_profit<0]
```

```python
loss_products = loss_products.sort_values() 
loss_products.head(10)
```

```python
loss_products.head(10).sort_values().plot(kind='bar')
plt.title('Top 10 loss making products')
plt.xlabel('Profit')
plt.ylabel('Products')
plt.show()
```
